```python
# importtant formulas to care of

# make columns names space free and in lowercase
df.columns = df.columns.str.lower()
df.columns = df.columns.str.replace(' ', '_')
df = df.rename(columns={'Purchase Amount (USD)':'purchase_amount_usd'})
```

```python
df.columns
```

```python
# creating a categorical column for age groups

labels = ['young_adult','adult','middle_aged','senior']

df['age_group'] = pd.qcut(df['age'], q = 4, labels = labels)
```

```python
# frequency Mapping

frequency = {
    'Fortnightly' : 14,
    'Monthly' : 30,
    'Annually' : 365,
    'Weekly' : 7,
    'Bi-Weekly' : 14,
    'Every 3 Months' : 90,
    'Quarterly' : 14
    
}

df['purchase_frequency_days'] = df['frequency_of_purchases'].map(frequency)
```
